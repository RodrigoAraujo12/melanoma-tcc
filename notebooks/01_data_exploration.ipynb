# Exploração do Dataset ISIC 2020
Analisa distribuição de classes, metadados clínicos e exemplos de imagens.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import os

# Caminhos no Kaggle
CSV_PATH = '/kaggle/input/siim-isic-melanoma-classification/train.csv'
IMAGES_DIR = '/kaggle/input/siim-isic-melanoma-classification/train'

df = pd.read_csv(CSV_PATH)
print(f'Total de amostras: {len(df)}')
df.head()
```

```python
# Distribuição de classes (desbalanceamento é esperado)
print(df['target'].value_counts())
df['target'].value_counts().plot(kind='bar', title='Distribuição: Benigno (0) vs Melanoma (1)')
plt.xticks([0, 1], ['Benigno', 'Melanoma'], rotation=0)
plt.ylabel('Quantidade')
plt.show()
```

```python
# Distribuição de idade e sexo
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df['age_approx'].hist(ax=axes[0], bins=20)
axes[0].set_title('Distribuição de Idade')
df['sex'].value_counts().plot(kind='bar', ax=axes[1], title='Distribuição de Sexo', rot=0)
plt.tight_layout()
plt.show()
```

```python
# Visualiza exemplos de imagens
melanoma_samples = df[df['target'] == 1].head(4)
benign_samples = df[df['target'] == 0].head(4)
samples = pd.concat([melanoma_samples, benign_samples])

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for ax, (_, row) in zip(axes.flat, samples.iterrows()):
    img_path = os.path.join(IMAGES_DIR, f"{row['image_name']}.jpg")
    img = Image.open(img_path)
    ax.imshow(img)
    label = 'Melanoma' if row['target'] == 1 else 'Benigno'
    ax.set_title(f"{label}\n{row['sex']}, {row['age_approx']}a")
    ax.axis('off')
plt.suptitle('Top: Melanoma | Bottom: Benigno')
plt.tight_layout()
plt.show()
```

```python
# Localização anatômica
df['anatom_site_general_challenge'].value_counts().plot(
    kind='barh', title='Localização das Lesões', figsize=(8, 5)
)
plt.tight_layout()
plt.show()
```
